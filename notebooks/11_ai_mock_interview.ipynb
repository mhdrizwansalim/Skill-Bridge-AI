```python
import random

def generate_interview(role, difficulty, technical_bank, hr_bank):

    technical_questions = technical_bank.get(role, {}).get(
        difficulty, []
    )

    hr_questions = hr_bank.get(difficulty, [])

    if not technical_questions:
        raise ValueError(
            f"No technical questions available for {role} at {difficulty} level."
        )

    # Select up to 5 technical questions
    selected_technical = random.sample(
        technical_questions,
        min(5, len(technical_questions))
    )

    # Select up to 2 HR questions
    selected_hr = random.sample(
        hr_questions,
        min(2, len(hr_questions))
    )

    interview_questions = []

    for question in selected_technical:
        interview_questions.append({
            "category": "Technical",
            "question": question
        })

    for question in selected_hr:
        interview_questions.append({
            "category": "HR",
            "question": question
        })

    random.shuffle(interview_questions)

    return interview_questions
```

```python
def generate_answer_feedback(row):

    score = row["relevance_score"]
    word_count = row["answer_length"]

    suggestions = []

    if score >= 40:
        suggestions.append("Good relevance to the question.")
    else:
        suggestions.append("Include more keywords related to the question.")

    if word_count < 10:
        suggestions.append("Provide a more detailed explanation.")

    elif word_count < 30:
        suggestions.append("Add examples to strengthen your answer.")

    else:
        suggestions.append("Keep your answer structured and concise.")

    suggestions.append(
        "Use the STAR method for behavioral questions."
        if row["category"] == "HR"
        else "Include practical examples and technical concepts."
    )

    return " ".join(suggestions)
```

```python
import json
import pandas as pd

answers_path = INTERVIEW_OUTPUT / "candidate_interview_answers.json"

with open(answers_path, "r", encoding="utf-8") as file:
    interview_answers = json.load(file)

print("Answers loaded:", len(interview_answers))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
def evaluate_answer(question, answer):

    question_words = set(
        re.findall(r"\b[a-zA-Z]{3,}\b", question.lower())
    )

    answer_words = set(
        re.findall(r"\b[a-zA-Z]{3,}\b", answer.lower())
    )

    if not answer.strip():
        return {
            "relevance_score": 0.0,
            "answer_length": 0,
            "feedback": "No answer provided."
        }

    overlap = question_words.intersection(answer_words)

    keyword_score = (
        len(overlap) / len(question_words) * 100
        if question_words else 0
    )

    try:
        vectorizer = TfidfVectorizer(stop_words="english")
        vectors = vectorizer.fit_transform([question, answer])

        similarity = cosine_similarity(
            vectors[0:1], vectors[1:2]
        )[0][0]

    except ValueError:
        similarity = 0

    score = (keyword_score * 0.6) + (similarity * 100 * 0.4)

    return {
        "relevance_score": round(float(score), 2),
        "answer_length": len(answer.split()),
        "feedback": "Keyword and text similarity evaluation completed."
    }
```

```python
evaluation_results = []

for item in interview_answers:

    result = evaluate_answer(
        item["question"],
        item["answer"]
    )

    evaluation_results.append({
        **item,
        **result
    })

evaluation_df = pd.DataFrame(evaluation_results)

display(
    evaluation_df[
        ["question", "relevance_score", "answer_length"]
    ]
)
```

```python
average_score = evaluation_df["relevance_score"].mean()

print("Average Interview Score:", round(average_score, 2))
```

```python
category_performance = (
    evaluation_df.groupby("category")["relevance_score"]
    .mean()
    .reset_index()
)

category_performance.columns = [
    "Category",
    "Average Score"
]

display(category_performance)
```

```python
def get_performance_level(score):

    if score >= 80:
        return "Excellent"
    elif score >= 60:
        return "Good"
    elif score >= 40:
        return "Average"
    else:
        return "Needs Improvement"


performance_level = get_performance_level(average_score)

print("Overall Score:", round(average_score, 2))
print("Performance Level:", performance_level)
```
