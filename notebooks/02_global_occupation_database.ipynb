```python
from pathlib import Path
from collections import Counter

def inspect_dataset(folder_path, dataset_name):

    files = [f for f in Path(folder_path).rglob("*") if f.is_file()]

    print(f"\n{'='*50}")
    print(f"{dataset_name} DATASET")
    print(f"{'='*50}")

    print("Total files:", len(files))

    extensions = Counter(
        file.suffix.lower() for file in files
    )

    print("\nFile types:")
    for extension, count in extensions.items():
        print(f"{extension}: {count}")

    print("\nTotal size (MB):",
          round(sum(f.stat().st_size for f in files) / (1024**2), 2))

    print("\nAll files:")
    for file in sorted(files):
        print(file.name)


inspect_dataset(ESCO_PATH, "ESCO")
inspect_dataset(ONET_PATH, "O*NET")
```

```python
# Initial dataset validation

datasets = {
    "ESCO Occupations": esco_occupations,
    "ESCO Skills": esco_skills,
    "ESCO Relationships": esco_relations,
    "O*NET Occupations": onet_occupations,
    "O*NET Skills": onet_skills,
    "O*NET Job Titles": onet_job_titles,
    "ESCO Skill Groups": esco_skill_groups
}

validation_results = []

for name, df in datasets.items():

    validation_results.append({
        "Dataset": name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Duplicate Rows": df.duplicated().sum(),
        "Missing Cells": df.isnull().sum().sum(),
        "Memory Usage (MB)": round(
            df.memory_usage(deep=True).sum() / (1024 ** 2), 2
        )
    })

validation_df = pd.DataFrame(validation_results)

display(validation_df)

print("\nTotal records across loaded datasets:",
      validation_df["Rows"].sum())

print("\nTotal duplicate rows:",
      validation_df["Duplicate Rows"].sum())

print("\nTotal missing cells:",
      validation_df["Missing Cells"].sum())
```

```python
# Save Day 2 validation report

REPORT_PATH = os.path.join(
    PROJECT_ROOT,
    "data",
    "processed",
    "day2_dataset_validation.csv"
)

os.makedirs(PROCESSED_DATA_PATH, exist_ok=True)

validation_df.to_csv(
    REPORT_PATH,
    index=False
)

print("Validation report saved successfully!")
print("Location:", REPORT_PATH)

# Final summary
print("\nSKILLBRIDGE DATASET SUMMARY")
print("=" * 40)
print("Loaded datasets:", len(datasets))
print("Total records:", validation_df["Rows"].sum())
print("Duplicate rows:", validation_df["Duplicate Rows"].sum())
print("Missing cells:", validation_df["Missing Cells"].sum())
```
