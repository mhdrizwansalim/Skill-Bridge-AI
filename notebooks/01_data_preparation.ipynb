```python
# Load datasets from CSV files

job_roles_df = pd.read_csv(
    os.path.join(RAW_DATA_PATH, "job_roles.csv")
)

skills_taxonomy_df = pd.read_csv(
    os.path.join(RAW_DATA_PATH, "skills_taxonomy.csv")
)

# Display dataset information

print("JOB ROLES DATASET")
print("========================")
print("Shape:", job_roles_df.shape)
display(job_roles_df.head())

print("\nSKILLS TAXONOMY DATASET")
print("========================")
print("Shape:", skills_taxonomy_df.shape)
display(skills_taxonomy_df.head())
```

```python
from collections import Counter

# Extract all required skills from job roles

all_required_skills = []

for skills in job_roles_df["required_skills"].dropna():
    skill_list = skills.split("|")
    all_required_skills.extend(
        [skill.strip() for skill in skill_list]
    )

# Count skill frequency

skill_frequency = Counter(all_required_skills)

# Convert into DataFrame

skill_frequency_df = pd.DataFrame(
    skill_frequency.items(),
    columns=["skill", "frequency"]
)

# Sort by frequency

skill_frequency_df = skill_frequency_df.sort_values(
    by="frequency",
    ascending=False
).reset_index(drop=True)

print("Total unique required skills:", len(skill_frequency_df))

display(skill_frequency_df.head(15))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.barplot(
    data=skill_frequency_df.head(15),
    x="frequency",
    y="skill"
)

plt.title("Top 15 Most Required Skills Across Job Roles")
plt.xlabel("Number of Job Roles")
plt.ylabel("Skills")
plt.tight_layout()
plt.show()
```

```python
# Create industry-wise skill mapping

industry_skills = {}

for _, row in job_roles_df.iterrows():

    industry = row["industry"]

    skills = row["required_skills"].split("|")

    if industry not in industry_skills:
        industry_skills[industry] = []

    industry_skills[industry].extend(
        [skill.strip() for skill in skills]
    )

# Display top 5 skills for each industry

for industry, skills in industry_skills.items():

    skill_counts = Counter(skills)

    print(f"\nIndustry: {industry}")
    print("-" * 40)

    for skill, count in skill_counts.most_common(5):
        print(f"{skill}: {count}")
```

```python
# Convert job roles into individual skill records

job_skill_records = []

for _, row in job_roles_df.iterrows():

    role = row["role"]
    industry = row["industry"]
    level = row["level"]

    skills = row["required_skills"].split("|")

    for skill in skills:

        job_skill_records.append({
            "role": role,
            "industry": industry,
            "level": level,
            "skill": skill.strip()
        })

# Create DataFrame

job_skill_df = pd.DataFrame(job_skill_records)

# Display results

print("Total job-skill records:", len(job_skill_df))

display(job_skill_df.head(15))
```

```python
job_skill_df.to_csv(
    os.path.join(PROCESSED_DATA_PATH, "job_skill_mapping.csv"),
    index=False
)

print("Job-skill mapping saved successfully!")
```

```python
# Normalize skill names for matching

job_skill_df["skill_key"] = (
    job_skill_df["skill"].str.strip().str.lower()
)

skills_taxonomy_df["skill_key"] = (
    skills_taxonomy_df["skill"].str.strip().str.lower()
)

# Map skill categories

job_skill_df = job_skill_df.merge(
    skills_taxonomy_df[["skill_key", "category"]],
    on="skill_key",
    how="left"
)

# Remove temporary column

job_skill_df.drop(columns=["skill_key"], inplace=True)

# Display results

print("Job-skill taxonomy mapping")
print("==========================")

print("Total records:", len(job_skill_df))
print("Unmapped skills:", job_skill_df["category"].isna().sum())

display(job_skill_df.head(15))
```

```python
job_skill_df.to_csv(
    os.path.join(PROCESSED_DATA_PATH, "job_skill_mapping.csv"),
    index=False
)

print("Updated job-skill mapping saved!")
```

```python
# Identify unmapped skills

unmapped_skills = (
    job_skill_df[
        job_skill_df["category"].isna()
    ]["skill"]
    .drop_duplicates()
    .sort_values()
)

print("Total unmapped skills:", len(unmapped_skills))

print("\nUnmapped skill names:")

for skill in unmapped_skills:
    print("-", skill)
```

```python
# Calculate taxonomy coverage

total_skills = len(job_skill_df)

mapped_skills = job_skill_df["category"].notna().sum()

unmapped_count = job_skill_df["category"].isna().sum()

coverage = (
    mapped_skills / total_skills * 100
    if total_skills > 0 else 0
)

print("Total job-skill records:", total_skills)
print("Mapped records:", mapped_skills)
print("Unmapped records:", unmapped_count)
print(f"Taxonomy coverage: {coverage:.2f}%")
```

```python
# Create industry-wise skill matrix

industry_skill_matrix = pd.crosstab(
    job_skill_df["industry"],
    job_skill_df["skill"]
)

print("Industry Skill Matrix Shape:")
print(industry_skill_matrix.shape)

display(industry_skill_matrix.head())
```

```python
# Visualize industry-wise skill distribution

plt.figure(figsize=(18, 10))

sns.heatmap(
    industry_skill_matrix,
    cmap="YlGnBu",
    linewidths=0.3
)

plt.title("Industry vs Required Skills")
plt.xlabel("Skills")
plt.ylabel("Industries")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
industry_skill_matrix.to_csv(
    os.path.join(
        PROCESSED_DATA_PATH,
        "industry_skill_matrix.csv"
    )
)

print("Industry skill matrix saved successfully!")
```

```python
# Calculate skill coverage for every job role

role_coverage = []

for role in job_roles_df["role"].unique():

    role_data = job_skill_df[
        job_skill_df["role"] == role
    ]

    total_skills = len(role_data)

    mapped_skills = role_data["category"].notna().sum()

    coverage = (
        mapped_skills / total_skills * 100
        if total_skills > 0 else 0
    )

    role_coverage.append({
        "role": role,
        "total_skills": total_skills,
        "mapped_skills": mapped_skills,
        "coverage_percentage": round(coverage, 2)
    })

# Create coverage DataFrame

role_coverage_df = pd.DataFrame(role_coverage)

# Sort by coverage

role_coverage_df = role_coverage_df.sort_values(
    by="coverage_percentage",
    ascending=True
).reset_index(drop=True)

display(role_coverage_df)
```

```python
plt.figure(figsize=(12, 9))

sns.barplot(
    data=role_coverage_df,
    x="coverage_percentage",
    y="role"
)

plt.title("Skill Taxonomy Coverage by Job Role")
plt.xlabel("Coverage Percentage")
plt.ylabel("Job Role")
plt.xlim(0, 100)

plt.tight_layout()
plt.show()
```

```python
role_coverage_df.to_csv(
    os.path.join(
        PROCESSED_DATA_PATH,
        "role_skill_coverage.csv"
    ),
    index=False
)

print("Role skill coverage report saved!")
```

```python
def generate_quality_report(df, dataset_name):

    report = pd.DataFrame({
        "column": df.columns,
        "data_type": df.dtypes.astype(str).values,
        "missing_values": df.isnull().sum().values,
        "missing_percentage": (
            df.isnull().mean().values * 100
        ).round(2),
        "unique_values": df.nunique().values
    })

    report.insert(0, "dataset", dataset_name)

    return report


# Generate reports

roles_quality = generate_quality_report(
    job_roles_df, "Job Roles"
)

skills_quality = generate_quality_report(
    skills_taxonomy_df, "Skills Taxonomy"
)

mapping_quality = generate_quality_report(
    job_skill_df, "Job Skill Mapping"
)

# Combine reports

quality_report = pd.concat(
    [
        roles_quality,
        skills_quality,
        mapping_quality
    ],
    ignore_index=True
)

display(quality_report)
```

```python
quality_report.to_csv(
    os.path.join(
        PROCESSED_DATA_PATH,
        "data_quality_report.csv"
    ),
    index=False
)

print("Data quality report generated successfully!")
```
