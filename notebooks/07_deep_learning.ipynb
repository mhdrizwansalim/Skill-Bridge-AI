```python
import pandas as pd
import numpy as np
import joblib

from sklearn.feature_extraction.text import TfidfVectorizer
from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent

ML_OUTPUT_PATH = PROJECT_ROOT / "data" / "processed" / "ml"

train_data = pd.read_csv(
    ML_OUTPUT_PATH / "ml_train.csv"
)

test_data = pd.read_csv(
    ML_OUTPUT_PATH / "ml_test.csv"
)

print("Training data:", train_data.shape)
print("Testing data:", test_data.shape)
```

```python
X_train_text = (
    train_data["occupationLabel"].astype(str)
    + " [SEP] "
    + train_data["skillLabel"].astype(str)
)

X_test_text = (
    test_data["occupationLabel"].astype(str)
    + " [SEP] "
    + test_data["skillLabel"].astype(str)
)

tfidf_vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    max_features=10000,
    min_df=2
)

X_train = tfidf_vectorizer.fit_transform(X_train_text)
X_test = tfidf_vectorizer.transform(X_test_text)

y_train = train_data["Has_Skill"]
y_test = test_data["Has_Skill"]

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
```

```python
X_train_dense = X_train.toarray().astype("float32")
X_test_dense = X_test.toarray().astype("float32")

print("Training tensor input:", X_train_dense.shape)
print("Testing tensor input:", X_test_dense.shape)
```

```python
import torch
import numpy as np

# Convert TF-IDF sparse matrices into dense arrays
X_train_dense = X_train.toarray().astype(np.float32)
X_test_dense = X_test.toarray().astype(np.float32)

# Convert features into PyTorch tensors
X_train_tensor = torch.tensor(X_train_dense, dtype=torch.float32)
X_test_tensor = torch.tensor(X_test_dense, dtype=torch.float32)

# Convert target labels into PyTorch tensors
y_train_tensor = torch.tensor(
    y_train.to_numpy(),
    dtype=torch.float32
).reshape(-1, 1)

y_test_tensor = torch.tensor(
    y_test.to_numpy(),
    dtype=torch.float32
).reshape(-1, 1)

print("X_train_tensor:", X_train_tensor.shape)
print("X_test_tensor:", X_test_tensor.shape)
print("y_train_tensor:", y_train_tensor.shape)
print("y_test_tensor:", y_test_tensor.shape)
```

```python
from torch.utils.data import TensorDataset, DataLoader

train_dataset = TensorDataset(
    X_train_tensor,
    y_train_tensor
)

test_dataset = TensorDataset(
    X_test_tensor,
    y_test_tensor
)

train_loader = DataLoader(
    train_dataset,
    batch_size=256,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=256,
    shuffle=False
)

print("Training batches:", len(train_loader))
print("Testing batches:", len(test_loader))
```

```python
import torch
import torch.nn as nn

class SkillBridgeANN(nn.Module):

    def __init__(self, input_features):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(input_features, 128),
            nn.ReLU(),

            nn.Linear(128, 64),
            nn.ReLU(),

            nn.Linear(64, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.network(x)


# Create model
input_features = X_train_tensor.shape[1]

model = SkillBridgeANN(input_features)

print(model)
```

```python
criterion = nn.BCELoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

print("Loss function:", criterion)
print("Optimizer:", optimizer)
```

```python
epochs = 10

for epoch in range(epochs):

    model.train()
    total_loss = 0

    for batch_X, batch_y in train_loader:

        optimizer.zero_grad()

        predictions = model(batch_X)

        loss = criterion(predictions, batch_y)

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    average_loss = total_loss / len(train_loader)

    print(
        f"Epoch [{epoch+1}/{epochs}], "
        f"Loss: {average_loss:.4f}"
    )
```

```python
class SkillBridgeLSTM(nn.Module):

    def __init__(self, vocab_size, embedding_dim, hidden_dim):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size,
            embedding_dim,
            padding_idx=0
        )

        self.lstm = nn.LSTM(
            embedding_dim,
            hidden_dim,
            batch_first=True
        )

        self.fc = nn.Linear(hidden_dim, 1)

    def forward(self, x):

        embedded = self.embedding(x)

        output, (hidden, cell) = self.lstm(embedded)

        final_output = hidden[-1]

        return torch.sigmoid(self.fc(final_output))


lstm_model = SkillBridgeLSTM(
    vocab_size=10000,
    embedding_dim=100,
    hidden_dim=128
)

print(lstm_model)
```

```python
class SkillBridgeBiLSTM(nn.Module):

    def __init__(self, vocab_size, embedding_dim, hidden_dim):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size,
            embedding_dim,
            padding_idx=0
        )

        self.lstm = nn.LSTM(
            embedding_dim,
            hidden_dim,
            batch_first=True,
            bidirectional=True
        )

        self.fc = nn.Linear(hidden_dim * 2, 1)

    def forward(self, x):

        embedded = self.embedding(x)

        output, (hidden, cell) = self.lstm(embedded)

        forward_hidden = hidden[-2]
        backward_hidden = hidden[-1]

        combined = torch.cat(
            (forward_hidden, backward_hidden),
            dim=1
        )

        return torch.sigmoid(self.fc(combined))


bilstm_model = SkillBridgeBiLSTM(
    vocab_size=10000,
    embedding_dim=100,
    hidden_dim=128
)

print(bilstm_model)
```

```python
import torch
import torch.nn as nn

class SkillBridgeCNN(nn.Module):

    def __init__(self, vocab_size, embedding_dim, num_filters):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size,
            embedding_dim,
            padding_idx=0
        )

        self.conv = nn.Conv1d(
            in_channels=embedding_dim,
            out_channels=num_filters,
            kernel_size=3
        )

        self.relu = nn.ReLU()
        self.pool = nn.AdaptiveMaxPool1d(1)

        self.fc = nn.Linear(num_filters, 1)

    def forward(self, x):

        embedded = self.embedding(x)

        embedded = embedded.transpose(1, 2)

        features = self.conv(embedded)
        features = self.relu(features)

        pooled = self.pool(features).squeeze(2)

        return torch.sigmoid(self.fc(pooled))


cnn_model = SkillBridgeCNN(
    vocab_size=10000,
    embedding_dim=100,
    num_filters=128
)

print(cnn_model)
```

```python
sample_input = torch.randint(
    1, 10000, (4, 20)
)

cnn_output = cnn_model(sample_input)

print("Input shape:", sample_input.shape)
print("CNN output shape:", cnn_output.shape)
print("CNN predictions:")
print(cnn_output)
```

```python
resume_skills = [
    "Python programming",
    "Building machine learning models",
    "Data visualization",
    "SQL database management",
    "Deep learning"
]

job_skills = [
    "Python",
    "Machine Learning",
    "Statistics",
    "Natural Language Processing",
    "SQL",
    "Data Analysis"
]
```

```python
resume_embeddings = embedding_model.encode(
    resume_skills,
    convert_to_tensor=True
)

job_embeddings = embedding_model.encode(
    job_skills,
    convert_to_tensor=True
)

print("Resume embeddings:", resume_embeddings.shape)
print("Job embeddings:", job_embeddings.shape)
```

```python
from sentence_transformers import util

similarity_matrix = util.cos_sim(
    resume_embeddings,
    job_embeddings
)

print("Similarity matrix shape:", similarity_matrix.shape)
print(similarity_matrix)
```

```python
import pandas as pd

results = []

for i, resume_skill in enumerate(resume_skills):

    best_index = similarity_matrix[i].argmax().item()

    results.append({
        "Resume Skill": resume_skill,
        "Best Matching Job Skill": job_skills[best_index],
        "Similarity Score": round(
            similarity_matrix[i][best_index].item(), 4
        )
    })

semantic_results = pd.DataFrame(results)

semantic_results
```
